```python
tokens = ["os", "produtos", "comprados", "não", "chegaram"]
stopwords = ["os", "não"]

tokens_limpos = []

for palavra in tokens:
    if palavra in stopwords:
        continue
    elif palavra == "comprados":
        tokens_limpos.append("comprar")
    elif palavra == "chegaram":
        tokens_limpos.append("chegar")
    else:
        tokens_limpos.append(palavra)

print(tokens_limpos)
```

```python
def validar_intencao(confianca):
    if confianca >= 0.70:
        return "Ação Executada"
    else:
        return "Fallback Acionado"


resultado = validar_intencao(0.55)
print(resultado)
```

```python
def calcular_peso_tfidf(frequencia_geral):
    if frequencia_geral == "muito_frequente":
        return 0.1
    elif frequencia_geral == "pouco_frequente":
        return 0.9


peso = calcular_peso_tfidf("pouco_frequente")
print(peso)
```

```python
def sao_sinonimos(distancia):
    if distancia < 0.20:
        return True
    else:
        return False


resultado = sao_sinonimos(0.05)
print(resultado)
```

```python
bruta = ["O", "sistema", "está", "muito", "lento", "hoje", "!"]
stopwords = ["O", "está"]

tokens_limpos = []

for palavra in bruta:
    if palavra not in stopwords and palavra != "!":
        tokens_limpos.append(palavra)

print(tokens_limpos)
```

```python
vp = 90
fn = 10

recall = vp / (vp + fn)

print(f"Recall: {recall * 100:.2f}%")
```

```python
probabilidades = [0.25, 0.30, 0.15]
limiar = 0.60

maior_confianca = max(probabilidades)

if maior_confianca >= limiar:
    print("Intenção Confirmada")
else:
    print("Acionar Fallback")
```
