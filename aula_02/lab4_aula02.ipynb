```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# 1. Dataset próprio
dados = {
    "mensagem": [
        "Quero comprar uma passagem",
        "Onde compro passagens?",
        "Passagens para São Paulo",
        "Quero reservar uma passagem",
        "Quero cancelar minha reserva",
        "Como faço pra cancelar?",
        "Não irei conseguir ir, quero cancelar",
        "Preciso cancelar minha passagem",
        "Como falo com o atendente?",
        "Quero falar com o suporte",
        "Onde é a central de atendimento ao cliente?",
        "Preciso falar com uma pessoa"
    ],
    "intencao": [
        "comprar_passagem",
        "comprar_passagem",
        "comprar_passagem",
        "comprar_passagem",
        "cancelar_reserva",
        "cancelar_reserva",
        "cancelar_reserva",
        "cancelar_reserva",
        "falar_atendente",
        "falar_atendente",
        "falar_atendente",
        "falar_atendente"
    ]
}

df = pd.DataFrame(dados)

# 2. Separar entrada (X) e resposta correta (y)
X = df["mensagem"]
y = df["intencao"]

# 3. Dividir em treino e teste
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# 4. Vetorizar os textos
vectorizer = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# 5. Criar e treinar o modelo
modelo = MultinomialNB()
modelo.fit(X_train_vec, y_train)

# Avaliação simples no conjunto de teste
predicoes_teste = modelo.predict(X_test_vec)
acuracia = accuracy_score(y_test, predicoes_teste)

print(f"Acurácia no conjunto de teste: {acuracia * 100:.2f}%")
print()

# 6. Frases inéditas
frases_novas = [
    "Me vê um bilhete pra São Paulo",
    "Não vou mais viajar",
    "Quero falar com o gerente"
]

# Transformar as frases inéditas usando o vocabulário já aprendido
frases_novas_vec = vectorizer.transform(frases_novas)

# 7. Fazer as previsões
predicoes_novas = modelo.predict(frases_novas_vec)

# 8. Exibir as intenções previstas no console
print("PREVISÕES PARA FRASES INÉDITAS")
print("-" * 40)

for frase, intencao in zip(frases_novas, predicoes_novas):
    print(f"Frase: {frase}")
    print(f"Intenção prevista: {intencao}")
    print()
```
